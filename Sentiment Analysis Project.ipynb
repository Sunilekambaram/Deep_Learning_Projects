# Sentiment Analysis Project - Fake Review dataset

```python
import nltk
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
```

# Step 1 - LOAD DATASET

```python
df=pd.read_csv("C:/Users/sunil/Downloads/Fake Reviews datasets.csv")
df
```

# Step 2 - SEPARATE WANTED COLUMN

```python
df1=df[['reviews']]
df1.head()
```

# Step 3 - CLEANING

```python
import re
def clean(text):
    text = re.sub('[^A-Za-z]+', ' ', text)
    return text

df1['Cleaned Reviews'] = df1['reviews'].apply(clean)
df1.head()
```

# Step 4 - LOWERCASE

```python
df1['Lowercase'] = df1['Cleaned Reviews'].apply(lambda x: " ".join(word.lower() for word in x.split()))
df1.head()
```

# Step 5 - STOPWORDS

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')
stop_words.remove("no")
stop_words.remove("not")
print(stop_words)
```

```python
df1['stopwords'] = df1['Lowercase'].apply(lambda x: " ".join(word for word in x.split() if word not in stop_words))
df1.head()
```

# Step 6 - TOKENIZATION AND POS

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet
from nltk import pos_tag

# Download necessary NLTK data files
nltk.download('punkt')
nltk.download('wordnet')    #WordNet is a large lexical database of English 

# POS tagger dictionary
pos_dict = {'J': wordnet.ADJ, 'V': wordnet.VERB, 'N': wordnet.NOUN, 'R': wordnet.ADV}

def token_pos(text):
    return [
        (word, pos_dict.get(tag[0]))
        for word, tag in pos_tag(word_tokenize(text))
    ]

# Apply to your data
df1['POS tagged'] = df1['stopwords'].apply(token_pos)
```

```python
df1.head()
```

# Step 7 - LEMMA

```python
from nltk.stem import WordNetLemmatizer

# Initialize the lemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

def lemmatize(pos_data):
    ans = " "
    for word, pos in pos_data:
        if not pos:  # If no POS tag is available, use the word as is
            lemma = word
        else:
            # Lemmatize based on the POS tag
            lemma = wordnet_lemmatizer.lemmatize(word, pos=pos)
        ans = ans + " " + lemma  # Add the lemma to the result string
    return ans

# Assuming df2 is your dataframe and 'POS tagged' contains POS-tagged text
df1['Lemma'] = df1['POS tagged'].apply(lemmatize)
```

```python
df1.head()
```

# Step 7 - ASSIGN SCORES FOR TESTING

```python
def scores(RATES):
    if RATES > 3:  # Positive threshold
        return 0   # positive
    elif RATES < 3:  # Negative threshold
        return 1 # negative
    else:
        return 2 # neutral
df["scores"]=df["rating"].apply(scores)
```

# Step 8 - SEPARATE X AND Y

```python
x=df1['Lemma']
y=df['scores']
```

```python
print(len(x))
print(len(y))
```

# Step 9 - TRAIN TEST SPLIT

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Step 10 - VECTORIZATION

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
x_train_vec = tfidf.fit_transform(x_train)
x_test_vec = tfidf.transform(x_test)
```

# Step 11 - MODEL

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
model = MultinomialNB()
model.fit(x_train_vec, y_train)
y_pred = model.predict(x_test_vec)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
